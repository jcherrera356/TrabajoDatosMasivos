# Prueba de rol: Admin

Este notebook lo corre la persona en el rol de **Admin** (quien creo el catalogo `datos_empresas`), como parte del Laboratorio 2b de `02_Gobernanza_Seguridad.ipynb`.

**Antes de correrlo, verificar:**
1. En el Account Console (User management > Groups) debe existir un grupo llamado exactamente `admin`, con esta cuenta como miembro. Las funciones de mascara del Laboratorio 2b chequean literalmente `is_account_group_member('admin')` -no el rol built-in de account admin de Databricks, sino un grupo con ese nombre exacto creado para este ejercicio.
2. Esta cuenta ya es `OWNER` del catalogo (lo creo), pero ser `OWNER` **no exime** de los column masks ni row filters en Unity Catalog -por eso el grupo `admin` es necesario incluso para quien creo todo.

**Que deberia ver:** todas las columnas con su valor real -documento del proveedor, valores del contrato, nombre de la empresa. Sin ningun dato enmascarado.

```python
%sql
-- Confirmar identidad y membresia del grupo antes de ver los datos
SELECT current_user() AS yo,
  is_account_group_member('admin') AS soy_admin,
  is_account_group_member('grupo_recursos_humanos') AS soy_rrhh,
  is_account_group_member('grupo_finanzas') AS soy_finanzas;
```

```python
%sql
-- Misma consulta de verificacion del Laboratorio 2b
SELECT id_contrato, proveedor_adjudicado, documento_proveedor_normalizado, sector, valor_contrato, valor_pagado
FROM datos_empresas.governance_lab.contratos_demo
LIMIT 10;
```

## Resultado esperado

`soy_admin = true`. En la segunda celda, **ninguna columna deberia salir en NULL**: documento, valor_contrato y valor_pagado con su valor real, igual que proveedor_adjudicado y sector.

Si alguna columna sale en NULL, revisar que la cuenta este realmente agregada al grupo `admin` en el Account Console -el cambio de membresia puede tardar unos minutos en propagarse.
