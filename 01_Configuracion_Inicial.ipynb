## Paso 1: Verificar y crear el catálogo y el esquema Bronze

```python
%sql
-- Ver qué catálogos están disponibles y en cuáles se tienen permisos
SHOW CATALOGS;
```

```python
%sql
-- Crear el catálogo del proyecto (idempotente)
CREATE CATALOG IF NOT EXISTS Datos_Empresas
COMMENT 'Catálogo del Trabajo Práctico 1 (Arquitectura y Gestión de Datos Masivos). Reúne datos sobre la salud del tejido empresarial de Colombia: el Registro Único Empresarial y Social (RUES) y la Tasa de Cambio Representativa del Mercado (TRM), como caso de estudio de una arquitectura Bronze/Silver/Gold con fuentes complementarias.';
```

```python
%sql
-- Crear el esquema Bronze siguiendo la arquitectura Medallion
-- Bronze: capa de datos crudos, tal como llegan de la fuente (inmutabilidad)
CREATE SCHEMA IF NOT EXISTS Datos_Empresas.bronze
COMMENT 'Capa Bronze - Datos crudos de RUES y TRM tal como se obtienen de sus APIs de origen (datos.gov.co), sin transformar nombres de columnas ni castear tipos de forma destructiva. Cada tabla incluye las columnas de auditoría _ingested_at y _source.';
```

```python
%sql
-- Verificar la creación
SHOW SCHEMAS IN Datos_Empresas;
```
