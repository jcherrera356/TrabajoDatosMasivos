# 01 - Configuración Silver: Esquemas Silver, DQL y Gold

Trabajo Práctico 2 - Arquitectura e Ingesta en Capa Bronze -> Silver -> Gold

Este notebook crea, dentro del mismo catálogo `Datos_Empresas` que ya existe del TP1, los tres esquemas nuevos que usan los demás notebooks de esta carpeta:

* **`silver`**: datos limpios, tipados y validados (RUES, SECOP, CIIU, TRM).
* **`dql`** (Data Quality Layer): registros rechazados por las reglas de calidad de Silver, con su motivo de rechazo.
* **`gold`**: tablas analíticas (dimensiones y hechos) listas para negocio.

Requiere que ya se haya ejecutado `01_Configuracion_Inicial.ipynb` y los notebooks de ingesta Bronze de la carpeta `Trabajo Practico 1 - Arquitectura Inicial (Capa Bronze)/`.

```python
%sql
-- Verificar que el catálogo del TP1 ya existe
SHOW SCHEMAS IN Datos_Empresas;
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS Datos_Empresas.silver
COMMENT 'Capa Silver - Datos limpios, tipados y validados de RUES, SECOP, CIIU y TRM. Las filas que incumplen alguna regla de calidad no llegan aqui, van al esquema dql.';

CREATE SCHEMA IF NOT EXISTS Datos_Empresas.dql
COMMENT 'Data Quality Layer - Registros rechazados durante el procesamiento Bronze -> Silver, con el motivo de rechazo (_motivo_rechazo) y la fecha de deteccion (_fecha_rechazo).';

CREATE SCHEMA IF NOT EXISTS Datos_Empresas.gold
COMMENT 'Capa Gold - Tablas analiticas (dimensiones y hechos) construidas a partir de Silver, listas para analisis de negocio.';

-- Verificar la creación
SHOW SCHEMAS IN Datos_Empresas;
```
